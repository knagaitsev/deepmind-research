```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load the CSV file
data = pd.read_csv("../results/ablation_100/all.csv")

# Convert time_index to string to ensure proper categorical handling
data["time_index"] = data["time_index"].astype(str)

# Create the plot
plt.figure(figsize=(8, 6))

# Group by time_index and plot each line separately
for time_index, group in data.groupby("time_index"):
    plt.plot(group["shell_index"], group["compared_output"], marker='o', linestyle='-', label=f"time_index {time_index}")

# Labels and title
plt.xlabel("Shell Index")
plt.ylabel("$|dy/y/\epsilon$|")
plt.title("Shell Index vs Compared Output for Different Time Indices")
plt.legend()
plt.grid(True)

# Ensure the output directory exists
output_dir = "../figs/ablation_100"
os.makedirs(output_dir, exist_ok=True)

# Save the figure
plt.savefig(os.path.join(output_dir, "all.pdf"))

plt.show()
```
