```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load the CSV file
data = pd.read_csv("../results/results_accuracy.csv")

# Create the output directory if it doesn't exist
output_dir = "../figs"
os.makedirs(output_dir, exist_ok=True)

# Plot the data
plt.figure(figsize=(8, 6))
plt.errorbar(data["time_index"], data["corr_mean"], yerr=data["corr_std"], fmt='-o', capsize=3, label="Correlation")
plt.xlabel("Time Index")
plt.ylabel("Correlation Mean")
plt.title("Correlation vs. Time Index")
plt.legend()
plt.grid(True)

plt.ylim(bottom=0)

# Save the figure as a PDF
output_path = os.path.join(output_dir, "accuracy.pdf")
plt.savefig(output_path, format="pdf")

print(f"Figure saved to {output_path}")

plt.show()
```
