```python
import matplotlib.pyplot as plt

# Shell edges data
shell_edges = [41, 975, 3296, 6901, 12023, 9476, 2706]

# Generate corresponding shell indices
shell_indices = list(range(0, len(shell_edges)))

# Create the plot
plt.figure(figsize=(8, 6))
plt.plot(shell_indices, shell_edges, marker='o', linestyle='-', color='gray')

# Add background grid
plt.grid(True, which='both', linestyle='--', linewidth=0.5, color='gray')

# Set plot labels and title
plt.xlabel('Shell Index')
plt.ylabel('Shell Edges')
plt.title('Shell Index vs Shell Edges')

# Save the plot to the specified file
plt.savefig('../figs/shell_edges.pdf')

# Show the plot (optional)
plt.show()
```
