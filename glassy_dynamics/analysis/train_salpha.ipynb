```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file into a pandas DataFrame
df = pd.read_csv('../results/train_salpha_stats.csv')

# Extract the relevant columns: 'step' and 'test_corr_mean'
steps = df['step']
test_corr_mean = df['test_corr_mean']

# Create the plot
plt.figure(figsize=(8, 6))
plt.plot(steps, test_corr_mean, label='Test Correlation Mean', color='b')

# Add labels and title
plt.xlabel('Training Step')
plt.ylabel('Test Correlation Mean')
plt.title('Step vs Test Correlation Mean')

# Optionally, add a grid for better readability
plt.grid(True)

# Save the plot as a PDF
plt.savefig('../figs/train_salpha.pdf')

# Show the plot (optional)
plt.show()
```
