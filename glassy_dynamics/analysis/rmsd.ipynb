```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data from the CSV file
df = pd.read_csv('../results/rmsd.csv')

# Remove non-numeric rows (like 'alpha')
df = df[pd.to_numeric(df['time_index'], errors='coerce').notnull()]

# Convert time_index and rmsd to appropriate data types
df['time_index'] = df['time_index'].astype(int)
df['rmsd'] = df['rmsd'].astype(float)

# Create the plot
plt.figure(figsize=(8, 6))
plt.plot(df['time_index'], df['rmsd'], marker='o', linestyle='-', color='b', label='RMSD')
plt.xlabel('Time Index')
plt.ylabel('RMSD')
plt.title('Time Index vs RMSD')

# Set y-axis to logarithmic scale
plt.yscale('log')

plt.grid(True)
plt.legend()

# Save the plot as a PDF
plt.savefig('../figs/rmsd.pdf')

# Show the plot
plt.show()
```
